```python
import pandas as pd

data = {
    "Size (microns)": [850, 600, 425, 300, 212, 150, 106, 75, 53, 38, 0],
    "Cumulative Passing (%)": [100.00, 97.50, 89.60, 71.20, 53.10, 37.80, 25.00, 15.30, 8.90, 4.40, 0.00]
}

df = pd.DataFrame(data)

# Calculate Weight Fraction
df["Cumulative Passing Fraction"] = df["Cumulative Passing (%)"] / 100
df["Weight Fraction"] = df["Cumulative Passing Fraction"].shift(fill_value=1) - df["Cumulative Passing Fraction"]

df.drop(columns=["Cumulative Passing Fraction","Cumulative Passing (%)"], inplace=True)

output_path = "sieve_data_output.csv"
df.to_csv(output_path, index=False)

df
```
